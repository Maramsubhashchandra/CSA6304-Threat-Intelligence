```python
def recommend_tool(task):
    tools = {
        "Wireshark": ["network", "packet", "traffic"],
        "Nmap": ["scan", "port", "host"],
        "Splunk": ["logs", "monitor", "siem"],
        "Nessus": ["vulnerability", "weakness", "security scan"],
        "Metasploit": ["penetration", "exploit", "attack"],
        "ClamAV": ["virus", "malware", "antivirus"]
    }

    task = task.lower()

    for tool, keywords in tools.items():
        if any(word in task for word in keywords):
            return tool

    return "No suitable tool found"


task = input("Enter security task: ")
print("Recommended Tool:", recommend_tool(task))
```
