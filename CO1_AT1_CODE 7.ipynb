```python

import re
from collections import Counter

logs = [
    "INFO Server started",
    "ERROR Login failed",
    "INFO User logged in",
    "ERROR Database failed",
    "WARNING High memory usage"
]

# 1. Logstash-style parsing
events = []
for log in logs:
    match = re.match(r"(\w+)\s+(.*)", log)
    if match:
        events.append({
            "level": match.group(1),
            "message": match.group(2)
        })

# 2. Elasticsearch-style indexing
index = {i: event for i, event in enumerate(events)}
print("Indexed Events:", len(index))

# 3. Kibana-style aggregation
counts = Counter(event["level"] for event in index.values())

print("\nKibana Dashboard:")
for level, count in counts.items():
    print(level, ":", count)
```
