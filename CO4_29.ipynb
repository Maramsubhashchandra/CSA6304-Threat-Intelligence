```python
# Incident containment recommendation

incident = input("Enter incident type: ").lower()

if incident == "malware":
    print("Strategy: Isolate infected systems.")
elif incident == "phishing":
    print("Strategy: Block malicious emails and reset passwords.")
elif incident == "ddos":
    print("Strategy: Block suspicious IPs and enable traffic filtering.")
elif incident == "data breach":
    print("Strategy: Restrict access and secure compromised accounts.")
elif incident == "ransomware":
    print("Strategy: Disconnect affected systems from the network.")
else:
    print("Strategy: Investigate the incident and monitor systems.")
```
