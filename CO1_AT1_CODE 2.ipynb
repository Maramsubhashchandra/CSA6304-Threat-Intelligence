```python
import socket

target = input("Enter target IP address: ")

print("Scanning target:", target)

for port in [21, 22, 23, 80, 443, 3306, 8080]:
    s = socket.socket()
    s.settimeout(1)
    result = s.connect_ex((target, port))

    if result == 0:
        print("Port", port, "is OPEN")

    s.close()

print("Scan completed.")
```
