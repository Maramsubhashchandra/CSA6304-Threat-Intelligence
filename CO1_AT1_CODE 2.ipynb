```python
import socket

host = input("Enter target host (e.g., 127.0.0.1): ")

for port in [21, 22, 80, 443, 3306]:
    s = socket.socket()
    s.settimeout(1)
    result = s.connect_ex((host, port))

    if result == 0:
        print("Port", port, "is OPEN")
    else:
        print("Port", port, "is CLOSED")

    s.close()
```
