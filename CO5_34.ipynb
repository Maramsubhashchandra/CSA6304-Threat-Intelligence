```python
data = [
    {"ip": "192.168.1.5", "login": 2, "status": "success"},
    {"ip": "10.0.0.8", "login": 15, "status": "failed"},
    {"ip": "45.10.20.1", "login": 20, "status": "failed"},
    {"ip": "192.168.1.9", "login": 3, "status": "success"}
]

# Known malicious IPs
threat_ips = ["45.10.20.1"]

# 1. Hypothesis-based hunting
print("Hypothesis-Based:")
for d in data:
    if d["login"] > 10:
        print(d)

# 2. Intelligence-based hunting
print("\nIntelligence-Based:")
for d in data:
    if d["ip"] in threat_ips:
        print(d)

# 3. Analytics-based hunting
print("\nAnalytics-Based:")
for d in data:
    if d["status"] == "failed" and d["login"] > 5:
        print(d)
```
