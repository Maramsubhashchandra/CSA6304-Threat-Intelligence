```python
from collections import Counter

# Sample web server logs (IP addresses)
logs = [
    "192.168.1.1",
    "192.168.1.2",
    "192.168.1.1",
    "192.168.1.1",
    "192.168.1.1",
    "192.168.1.3",
    "192.168.1.1"
]

threshold = 3

count = Counter(logs)

print("DoS Attack Detection:")

for ip, requests in count.items():
    if requests > threshold:
        print("Possible DoS Attack from:", ip,
              "| Requests:", requests)
    else:
        print("Normal IP:", ip, "| Requests:", requests)
```
