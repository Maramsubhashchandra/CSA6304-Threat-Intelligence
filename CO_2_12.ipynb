```python
alert = input("Enter SOC alert description: ").lower()

if "ransomware" in alert or "data breach" in alert:
    priority = "P1 - Critical"
elif "malware" in alert or "unauthorized access" in alert:
    priority = "P2 - High"
elif "failed login" in alert or "suspicious activity" in alert:
    priority = "P3 - Medium"
else:
    priority = "P4 - Low"

print("Alert Priority:", priority)
```
