```python
# Signature-based IDS

signatures = [
    {"protocol": "TCP", "port": 80, "pattern": "SQL_INJECTION"},
    {"protocol": "TCP", "port": 443, "pattern": "MALICIOUS_PAYLOAD"}
]

traffic = [
    {"protocol": "TCP", "port": 80, "data": "SQL_INJECTION"},
    {"protocol": "UDP", "port": 80, "data": "SQL_INJECTION"},
    {"protocol": "TCP", "port": 443, "data": "MALICIOUS_PAYLOAD"}
]

for packet in traffic:
    alert = False

    for sig in signatures:
        if (packet["protocol"] == sig["protocol"] and
            packet["port"] == sig["port"] and
            sig["pattern"] in packet["data"]):
            alert = True
            break

    if alert:
        print("ALERT: Intrusion Detected!", packet)
    else:
        print("Safe Traffic:", packet)
```
