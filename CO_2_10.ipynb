```python
connections = set()

n = int(input("Enter number of packets: "))

for i in range(n):
    direction = input("Enter direction (outbound/inbound): ").lower()
    src = input("Enter source IP: ")
    dst = input("Enter destination IP: ")

    if direction == "outbound":
        connections.add((src, dst))
        print("Packet Allowed - Outbound connection recorded")

    elif direction == "inbound":
        if (dst, src) in connections:
            print("Packet Allowed - Return traffic")
        else:
            print("Packet Blocked - Unsolicited inbound traffic")

    else:
        print("Invalid direction")
```
