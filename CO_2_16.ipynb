```python
layers = ["Firewall", "IDS", "Antivirus", "Access Control"]
blocked = False

for layer in layers:
    result = input(f"Did {layer} block the attack? (yes/no): ").lower()
    if result == "yes":
        print(layer, "blocked the attack!")
        blocked = True
        break
    else:
        print(layer, "failed to block the attack.")

if blocked:
    print("Result: System Protected")
else:
    print("Result: Attack Reached the Target!")
```
