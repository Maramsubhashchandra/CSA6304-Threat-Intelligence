```python
users = {
    "Harish": ("Admin", ["read", "write", "delete"]),
    "Ravi": ("Employee", ["read", "write", "delete"]),
    "Anitha": ("Guest", ["read", "write"])
}

roles = {
    "Admin": ["read", "write", "delete"],
    "Employee": ["read", "write"],
    "Guest": ["read"]
}

for name, (role, permissions) in users.items():
    excess = set(permissions) - set(roles[role])

    if excess:
        print(name, "Excess Access:", excess)
    else:
        print(name, "Access OK")
```
