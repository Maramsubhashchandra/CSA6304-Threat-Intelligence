```python
report = input("Enter post-incident report: ").lower()

areas = {
    "Patch Management": ["patch", "update", "vulnerability"],
    "Security Awareness": ["training", "awareness", "phishing"],
    "Monitoring": ["monitoring", "logs", "alerts"],
    "Backup Strategy": ["backup", "recovery", "restore"],
    "Access Control": ["access", "permission", "authentication"]
}

print("Improvement Areas Identified:")

found = False
for area, keywords in areas.items():
    if any(word in report for word in keywords):
        print("-", area)
        found = True

if not found:
    print("No improvement areas identified.")
```
