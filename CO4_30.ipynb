```python
# SIEM Log Correlation

firewall = ["192.168.1.10", "192.168.1.10"]
windows = ["192.168.1.10", "192.168.1.10", "192.168.1.10"]
vpn = ["192.168.1.10"]

for ip in set(firewall):
    if ip in windows and ip in vpn:
        if windows.count(ip) >= 3:
            print("ALERT: Possible Brute-Force Attack!")
            print("Attacker IP:", ip)
            print("Action: Block IP and investigate.")
        else:
            print("Normal Activity")
```
