```python
def classify_control(description):
    domains = {
        "Organizational Controls": ["policy", "risk", "management", "supplier", "compliance"],
        "People Controls": ["employee", "training", "awareness", "staff", "personnel"],
        "Physical Controls": ["door", "lock", "CCTV", "building", "access card"],
        "Technological Controls": ["firewall", "password", "encryption", "antivirus", "network"]
    }

    description = description.lower()

    for domain, keywords in domains.items():
        for word in keywords:
            if word.lower() in description:
                return domain

    return "Control domain not identified"


control = input("Enter security control description: ")
print("ISO 27001 Annex A Domain:", classify_control(control))
```
