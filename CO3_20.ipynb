```python
def classify_threat(description):
    keywords = {
        "Phishing": ["email", "fake website", "password", "link"],
        "DoS/DDoS": ["flood", "traffic", "server", "overload"],
        "Malware": ["virus", "trojan", "ransomware", "malware"],
        "Man-in-the-Middle": ["intercept", "communication", "eavesdrop", "interception"]
    }

    for threat, words in keywords.items():
        for word in words:
            if word in description.lower():
                return threat

    return "Unknown Threat"

description = input("Enter incident description: ")
print("Threat Category:", classify_threat(description))
```
