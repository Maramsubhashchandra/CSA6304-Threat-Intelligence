```python
# Simple Firewall Rule Engine

rules = [
    {"protocol": "TCP", "port": 80, "action": "ALLOW"},
    {"protocol": "TCP", "port": 443, "action": "ALLOW"},
    {"protocol": "UDP", "port": 53, "action": "ALLOW"},
    {"protocol": "ANY", "port": "ANY", "action": "DENY"}
]

packets = [
    {"protocol": "TCP", "port": 80},
    {"protocol": "UDP", "port": 53},
    {"protocol": "TCP", "port": 22}
]

for packet in packets:
    result = "DENY"

    for rule in rules:
        if (rule["protocol"] == packet["protocol"] and
            rule["port"] == packet["port"]):
            result = rule["action"]
            break

        if rule["protocol"] == "ANY":
            result = rule["action"]
            break

    print(packet, "->", result)
```
