```python
text = input("Enter incident description: ").lower()

if "phishing" in text or "fake email" in text:
    cause = "Phishing"
elif "unpatched" in text or "outdated software" in text:
    cause = "Unpatched Vulnerability"
elif "misconfiguration" in text or "wrong settings" in text:
    cause = "Misconfiguration"
elif "employee" in text or "insider" in text:
    cause = "Insider Threat"
elif "third-party" in text or "supplier" in text:
    cause = "Supply-Chain Compromise"
else:
    cause = "Unknown"

print("Root Cause:", cause)
```
