```python
from urllib.parse import urlparse
import ipaddress

url = input("Enter URL: ").lower()
parsed = urlparse(url if "://" in url else "http://" + url)
host = parsed.hostname or ""

score = 0
keywords = ["login", "verify", "update", "secure", "bank", "free"]

if any(word in url for word in keywords):
    score += 1

try:
    ipaddress.ip_address(host)
    score += 1
except ValueError:
    pass

if len(host.split(".")) > 3:
    score += 1

if "@" in parsed.netloc:
    score += 1

if score >= 2:
    print("Result: Suspicious URL - Possible Phishing")
else:
    print("Result: URL appears less suspicious")
```
