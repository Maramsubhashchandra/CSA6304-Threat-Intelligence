```python
# Malicious traffic signatures
signatures = ["malware", "sql injection", "ddos"]

# Trusted IP addresses
whitelist = ["192.168.1.10"]

traffic = [
    ("192.168.1.10", "malware"),
    ("192.168.1.20", "sql injection"),
    ("192.168.1.30", "normal traffic")
]

for ip, data in traffic:
    if ip in whitelist:
        print(ip, "Allowed (Trusted Source)")
    elif any(sig in data.lower() for sig in signatures):
        print(ip, "Blocked (Malicious Traffic)")
    else:
        print(ip, "Allowed (Safe Traffic)")
```
