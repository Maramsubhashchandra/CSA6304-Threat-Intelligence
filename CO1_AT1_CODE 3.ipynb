```python
policy = input("Enter security policy document: ").lower()

elements = {
    "Scope": "scope",
    "Roles and Responsibilities": "roles",
    "Enforcement": "enforcement",
    "Review Cycle": "review",
    "Approval Authority": "approval"
}

missing = []

for name, keyword in elements.items():
    if keyword not in policy:
        missing.append(name)

if missing:
    print("\nMissing Governance Elements:")
    for item in missing:
        print("-", item)
else:
    print("\nAll required governance elements are present.")
```
