```python
report = input("Enter threat report description: ").lower()
audience = input("Enter intended audience: ").lower()

if any(word in report for word in ["business risk", "long-term", "trend", "executive"]):
    print("Classification: Strategic")
elif any(word in report for word in ["tactics", "techniques", "procedures", "mitre", "ioc"]):
    print("Classification: Tactical")
elif any(word in report for word in ["ip address", "malware sample", "active attack", "incident", "alert"]):
    print("Classification: Operational")
else:
    print("Classification: Operational")
```
