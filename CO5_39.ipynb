```python
feed = [" 192.168.1.10 ", "MALWARE.COM", "10.0.0.5"]

logs = ["192.168.1.10", "safe.com", "10.0.0.8"]

# Normalize IOCs
iocs = [x.strip().lower() for x in feed]

print("Threat Intelligence Alerts:")

for log in logs:
    if log.strip().lower() in iocs:
        print("ALERT: IOC Match Found -", log)
```
