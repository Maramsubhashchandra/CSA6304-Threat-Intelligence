```python
logs = [
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "success"),
    ("john", "success")
]

user = input("Enter username: ")

failed = 0
detected = False

for name, status in logs:
    if name == user:
        if status == "failed":
            failed += 1
        elif status == "success" and failed >= 3:
            detected = True

if detected:
    print("Brute-force attack detected!")
else:
    print("No brute-force attack detected.")
```
