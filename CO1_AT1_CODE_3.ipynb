```python
policy = input("Enter security policy document: ").lower()

elements = ["scope", "roles and responsibilities",
            "enforcement", "review cycle", "approval authority"]

missing = [item for item in elements if item not in policy]

if missing:
    print("Missing governance elements:")
    for item in missing:
        print("-", item)
else:
    print("All governance elements are present.")
```
