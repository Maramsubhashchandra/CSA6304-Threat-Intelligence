```python
events = [
    {"user": "John", "action": "login", "status": "success"},
    {"user": "Alice", "action": "login", "status": "failed"},
    {"user": "John", "action": "logout", "status": "success"},
    {"user": "Bob", "action": "login", "status": "failed"},
    {"user": "Alice", "action": "login", "status": "success"}
]

# Search filter
print("Failed Login Events:")
for e in events:
    if e["action"] == "login" and e["status"] == "failed":
        print(e)

# Stats count by user
print("\nEvent Count By User:")
count = {}

for e in events:
    count[e["user"]] = count.get(e["user"], 0) + 1

for user, total in count.items():
    print(user, ":", total)
```
