```python
# CIA Security Violation Detection

properties = [
    ("Data Encryption", "Confidentiality", True),
    ("Access Control", "Confidentiality", False),
    ("File Integrity", "Integrity", True),
    ("Backup System", "Availability", False)
]

violations = set()

for name, cia, status in properties:
    if not status:
        violations.add(cia)

print("CIA Violations:")

if violations:
    for cia in violations:
        print(cia, "is Violated")
else:
    print("No CIA Violations")
```
