```python
import hashlib

# Malware sample (simulated)
sample = b"CreateFile VirtualAlloc WriteProcessMemory"

# Static Analysis
print("STATIC ANALYSIS")
print("SHA-256:", hashlib.sha256(sample).hexdigest())

apis = ["VirtualAlloc", "WriteProcessMemory", "CreateFile"]
for api in apis:
    if api.encode() in sample:
        print("Suspicious API found:", api)

# Dynamic Analysis (simulated behavior log)
print("\nDYNAMIC ANALYSIS")
logs = ["File created", "Network connection", "Registry modified"]

suspicious = ["Network connection", "Registry modified"]

for log in logs:
    if log in suspicious:
        print("Suspicious behavior:", log)

print("\nMalware Analysis Completed")
```
