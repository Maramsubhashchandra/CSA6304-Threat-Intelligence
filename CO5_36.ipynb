```python
def identify_compliance(description):
    d = description.lower()

    if any(x in d for x in ["credit card", "payment", "cardholder"]):
        return "PCI-DSS"
    elif any(x in d for x in ["patient", "medical", "healthcare"]):
        return "HIPAA"
    elif any(x in d for x in ["personal data", "privacy", "eu citizen"]):
        return "GDPR"
    elif any(x in d for x in ["financial report", "accounting", "audit"]):
        return "SOX"
    else:
        return "No matching standard found"

description = input("Enter system/data description: ")
print("Applicable Standard:", identify_compliance(description))
```
