```python
import hashlib

records = [
    {"data": "Evidence1", "time": 100, "prev": "0"},
    {"data": "Evidence2", "time": 200, "prev": ""},
    {"data": "Evidence3", "time": 300, "prev": ""}
]

prev_hash = "0"
prev_time = 0

for r in records:
    current_hash = hashlib.sha256(
        (r["data"] + str(r["time"]) + prev_hash).encode()
    ).hexdigest()

    if r["prev"] != prev_hash or r["time"] < prev_time:
        print("Chain Invalid")
        break

    print("Record Verified:", r["data"])
    prev_hash = current_hash
    prev_time = r["time"]
else:
    print("Chain of Custody Verified Successfully")
```
