```python
activity = input("Enter intelligence team activity: ").lower()

if "requirement" in activity or "priority" in activity:
    stage = "Direction"
elif "gather" in activity or "collect" in activity:
    stage = "Collection"
elif "clean" in activity or "format" in activity:
    stage = "Processing"
elif "analyze" in activity or "investigate" in activity:
    stage = "Analysis"
elif "report" in activity or "share" in activity:
    stage = "Dissemination"
elif "feedback" in activity or "improve" in activity:
    stage = "Feedback"
else:
    stage = "Unknown Stage"

print("Threat Intelligence Lifecycle Stage:", stage)
```
