```python
from scapy.all import IP, TCP, UDP, Ether

count = int(input("Enter number of packets to capture: "))

packets = [
    Ether()/IP(src="192.168.1.2", dst="8.8.8.8")/TCP(),
    Ether()/IP(src="192.168.1.3", dst="8.8.4.4")/UDP(),
    Ether()/IP(src="10.0.0.1", dst="192.168.1.1")/TCP()
]

print("Packet Summaries:")
for packet in packets[:count]:
    print(packet.summary())
```
