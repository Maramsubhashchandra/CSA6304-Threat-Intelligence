```python

# Network segmentation access policy

policies = {
    "Low-Trust": ["Internet"],
    "Corporate": ["Internet", "Low-Trust"],
    "Sensitive": ["Corporate"]
}

source = input("Enter source segment: ")
target = input("Enter target segment: ")

print("\nSource:", source)
print("Target:", target)

if target in policies.get(source, []):
    print("Access Granted")
else:
    print("Access Denied - Lateral Movement Blocked")
```
