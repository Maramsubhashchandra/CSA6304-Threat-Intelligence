```python
scenario = input("Enter business scenario: ").lower()

if "bank" in scenario or "transaction" in scenario or "payment" in scenario:
    print("Application Area: Banking Fraud Prevention")
elif "hospital" in scenario or "patient" in scenario or "medical" in scenario:
    print("Application Area: Healthcare Data Protection")
elif "shopping" in scenario or "online store" in scenario or "customer order" in scenario:
    print("Application Area: E-commerce Fraud Prevention")
else:
    print("Application Area: General Cyber Threat Intelligence")
```
