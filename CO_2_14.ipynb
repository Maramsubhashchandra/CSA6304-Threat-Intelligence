```python
events = [
    ("Firewall", "192.168.1.10", "Failed Login"),
    ("Server", "192.168.1.10", "Failed Login"),
    ("IDS", "192.168.1.10", "Malicious Activity")
]

print("1. Collection: Events collected =", len(events))

# Normalize events
data = [(s, ip, e.lower()) for s, ip, e in events]
print("2. Normalization: Events standardized")

# Correlate events by IP
count = {}
for s, ip, e in data:
    count[ip] = count.get(ip, 0) + 1

# Generate alert
print("3. Correlation and Alert:")
for ip, n in count.items():
    if n >= 3:
        print("ALERT: Suspicious activity from", ip)
    else:
        print("No alert for", ip)
```
