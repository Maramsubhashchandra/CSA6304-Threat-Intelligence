```python
baseline = {
    "max_attempts": 3,
    "allowed_location": "India"
}

events = [
    {"user": "Harish", "attempts": 2, "location": "India"},
    {"user": "Ravi", "attempts": 8, "location": "USA"},
    {"user": "Anitha", "attempts": 1, "location": "India"}
]

print("Continuous Security Monitoring Started")

for event in events:
    if (event["attempts"] > baseline["max_attempts"] or
        event["location"] != baseline["allowed_location"]):
        print("ALERT:", event["user"], "- Anomaly Detected")
    else:
        print(event["user"], "- Normal Login")

print("Monitoring Completed")
```
